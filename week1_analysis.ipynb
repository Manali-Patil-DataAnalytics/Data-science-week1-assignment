```python
import pandas as pd

clean_df = pd.read_csv("../Data/cleaned/ecommerce_sales_cleaned.csv")

clean_df["order_date"] = pd.to_datetime(clean_df["order_date"])

print(clean_df.shape)
```

```python
total_revenue = clean_df["revenue"].sum()

print("Total Revenue:", round(total_revenue, 2))
```

```python
total_orders = clean_df["order_id"].nunique()

print("Total Orders:", total_orders)
```

```python
average_order_value = total_revenue / total_orders

print("Average Order Value:", round(average_order_value, 2))
```

```python

total_units_sold = clean_df["quantity"].sum()

print("Total Units Sold:", total_units_sold)
```

```python

category_sales = clean_df.groupby("product_category")["quantity"].sum()

category_sales = category_sales.sort_values(ascending=False)

print("Units Sold by Category:")
print(category_sales)

print("\nTop-Selling Category:", category_sales.idxmax())
```

```python

category_revenue = clean_df.groupby("product_category")["revenue"].sum()

category_revenue = category_revenue.sort_values(ascending=False)

print("Revenue by Category:")
print(category_revenue)

print("\nTop Revenue-Generating Category:", category_revenue.idxmax())
```

```python

clean_df["month"] = clean_df["order_date"].dt.to_period("M")

monthly_sales = clean_df.groupby("month")["revenue"].sum()

print("Monthly Revenue:")
print(monthly_sales)
```

```python

print("First Order Date:", clean_df["order_date"].min())
print("Last Order Date:", clean_df["order_date"].max())
print("Number of Months:", clean_df["month"].nunique())
```

```python

highest_month = monthly_sales.idxmax()
highest_revenue = monthly_sales.max()

lowest_month = monthly_sales.idxmin()
lowest_revenue = monthly_sales.min()

print("Highest Revenue Month:", highest_month)
print("Highest Monthly Revenue:", round(highest_revenue, 2))

print("\nLowest Revenue Month:", lowest_month)
print("Lowest Monthly Revenue:", round(lowest_revenue, 2))
```

```python

monthly_category_sales = clean_df.pivot_table(
    index="month",
    columns="product_category",
    values="revenue",
    aggfunc="sum"
)

print(monthly_category_sales.head(10))
```

```python

import matplotlib.pyplot as plt

monthly_category_sales.plot(
    kind="line",
    figsize=(14, 6),
    marker="o"
)

plt.title("Monthly Revenue Trend by Product Category")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.legend(title="Product Category")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
print("Top Revenue Category for Each Month:")
print(monthly_category_sales.idxmax(axis=1).head(10))
```

```python
region_revenue = clean_df.groupby("region")["revenue"].sum()

region_revenue = region_revenue.sort_values(ascending=False)

print("Revenue by Region:")
print(region_revenue)
```

```python
import matplotlib.pyplot as plt

region_revenue.plot(kind="bar", figsize=(8, 5))

plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
customer_order_counts = (
    clean_df[clean_df["customer_id"] != "Unknown"]
    .groupby("customer_id")["order_id"]
    .nunique()
)

repeat_customers = customer_order_counts[customer_order_counts > 1]

print("Total Identified Customers:", len(customer_order_counts))
print("Repeat Customers:", len(repeat_customers))
print("One-Time Customers:", (customer_order_counts == 1).sum())
```

```python
payment_counts = clean_df["payment_method"].value_counts()

print("Orders by Payment Method:")
print(payment_counts)
```

```python
import matplotlib.pyplot as plt

payment_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7)
)

plt.title("Payment Method Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
total_revenue = clean_df["revenue"].sum()

total_orders = clean_df["order_id"].nunique()

average_order_value = total_revenue / total_orders

total_units_sold = clean_df["quantity"].sum()

top_selling_category = (
    clean_df.groupby("product_category")["quantity"]
    .sum()
    .idxmax()
)

top_revenue_category = (
    clean_df.groupby("product_category")["revenue"]
    .sum()
    .idxmax()
)

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Orders",
        "Average Order Value",
        "Total Units Sold",
        "Top-Selling Category",
        "Top Revenue-Generating Category"
    ],
    "Value": [
        round(total_revenue, 2),
        total_orders,
        round(average_order_value, 2),
        int(total_units_sold),
        top_selling_category,
        top_revenue_category
    ]
})

print(kpi_summary)
```

```python

import pandas as pd

df = pd.read_csv("../Data/cleaned/ecommerce_sales_cleaned.csv")
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")


print("1. REVENUE BY CATEGORY")
print(df.groupby("product_category")["revenue"].sum().sort_values(ascending=False).round(2))

print("\n2. UNITS SOLD BY CATEGORY")
print(df.groupby("product_category")["quantity"].sum().sort_values(ascending=False))

print("\n3. REVENUE BY REGION")
print(df.groupby("region")["revenue"].sum().sort_values(ascending=False).round(2))

print("\n4. MONTHLY REVENUE")
monthly_revenue = (
    df.dropna(subset=["order_date"])
      .groupby(df.dropna(subset=["order_date"])["order_date"].dt.to_period("M"))["revenue"]
      .sum()
      .sort_index()
)
print("Highest revenue month:", monthly_revenue.idxmax(), round(monthly_revenue.max(), 2))
print("Lowest revenue month:", monthly_revenue.idxmin(), round(monthly_revenue.min(), 2))

# 5. Repeat customer analysis (exclude unknown customer IDs)
known_customers = df[df["customer_id"].astype(str).str.lower() != "unknown"]
customer_orders = known_customers.groupby("customer_id")["order_id"].nunique()
repeat_customers = (customer_orders > 1).sum()
one_time_customers = (customer_orders == 1).sum()
total_known_customers = len(customer_orders)

print("\n5. CUSTOMER PURCHASE FREQUENCY")
print("Known customers:", total_known_customers)
print("Repeat customers:", repeat_customers)
print("One-time customers:", one_time_customers)
if total_known_customers:
    print("Repeat customer rate:", round(repeat_customers / total_known_customers * 100, 2), "%")

print("\n6. PAYMENT METHOD DISTRIBUTION")
payment_methods = (
    df["payment_method"].fillna("Unknown").astype(str)
      .str.strip().str.lower().str.title()
      .value_counts()
)
print(payment_methods)


print("\n7. DELIVERY TIME")
print("Average delivery days:", round(df["delivery_days"].mean(), 2))
print("Median delivery days:", round(df["delivery_days"].median(), 2))


print("\n8. DATASET DATE RANGE")
print("First order date:", df["order_date"].min())
print("Last order date:", df["order_date"].max())
```

 Key Findings and Observations

1. Total Revenue: The dataset generated total revenue of ₹5,107,881.09 from 4,996 orders.

2. Top-Performing Category: Electronics generated the highest revenue of ₹1,830,224.57 and recorded the highest sales quantity of 7,112 units.

3. Regional Performance:The West region generated the highest revenue of ₹1,346,889.82, while the East region generated the lowest revenue of ₹1,234,232.40.

4. Monthly Revenue Trend: August 2026 recorded the highest monthly revenue of ₹44,728.82. September 2035 recorded the lowest revenue of ₹4,952.23. However, the September 2035 data is incomplete because the last order date is September 9, 2035.

5. Customer Purchase Behaviour: Among 989 customers with known customer IDs, 950 were repeat customers and 39 were one-time customers. This indicates frequent repeat purchasing in the dataset, although the unusually high repeat-customer rate should be checked before drawing a general conclusion.

6. Payment Method: Card was the most frequently recorded payment method, with 2,270 orders. COD accounted for 1,772 orders, while Wallet accounted for 954 orders.

7. Delivery Time: The average delivery time was 6.12 days, while the median delivery time was 6 days.

8. Dataset Limitation: The dataset does not contain product names or city names. Therefore, individual top-selling products and city-level sales performance cannot be identified.

9. Date Range Limitation: The dataset covers orders from January 2022 to September 2035. This long date range should be checked against the dataset's source information before interpreting long-term trends.
